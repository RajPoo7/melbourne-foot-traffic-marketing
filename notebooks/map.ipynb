```python
from pathlib import Path
import pandas as pd
import geopandas as gpd  # this is in your env

# 1. Set up paths (assuming you are in the `notebooks/` folder)
REPO = Path("..").resolve()
RAW = REPO / "data" / "raw"
OUT = REPO / "analytics" / "looker_studio_datasources"

# 2. Read sensor lookup (id + name)
lookup = pd.read_csv(OUT / "sensors_lookup.csv")

# 3. Read locations (name + latitude/longitude) from the GeoJSON
locs = gpd.read_file(RAW / "sensor_locations.geojson")

# 4. Keep only the bits we need from the locations file
locs_small = locs[["sensor_name", "latitude", "longitude"]]

# 5. Join them together on sensor_name
sensor_map = lookup.merge(locs_small, on="sensor_name", how="left")

# Quick sanity check: did everyone get a coordinate?
print("Rows:", len(sensor_map))
print("Rows missing coords:", sensor_map["latitude"].isna().sum())

# 6. Save as a new CSV for Looker Studio
out_path = OUT / "sensor_locations.csv"
sensor_map.to_csv(out_path, index=False)
print("Wrote:", out_path.relative_to(REPO))

sensor_map.head()
```
